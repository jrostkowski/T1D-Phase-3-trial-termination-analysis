```python
import requests
import json
import os
from datetime import date
```

```python
#Define the API endpoint and parameters
url = "https://clinicaltrials.gov/api/v2/studies"
parameters = {
    'format':'json',
    'query.cond':'type 1 diabetes',
    'filter.advanced':'AREA[Phase]PHASE3',
    'countTotal': 'true',
    'pageSize': 100,
    }


#Define a list to store all trials and a variable for the page token
all_trials = []
pageToken = None
counter = 0

#Loop through the API requests until there are no more pages of results
while counter < 20:
    if pageToken:
        parameters['pageToken'] = pageToken
    get_api = requests.get(url, params=parameters)
    if get_api.status_code != 200:                   #Check if the API request was successful
        print(f"Error: {get_api.status_code}")
        break
    data = get_api.json()
    all_trials.extend(data['studies'])
    print(len(all_trials))
    counter = counter + 1
    if data.get('nextPageToken') is None:
        break
    else:
        pageToken = data['nextPageToken']
    if 'totalCount' in data:
        total_count = data['totalCount']         #Check if the total count is present in the response and store it in a variable

print(len(all_trials), total_count)
```

```python
#Check for duplicates in the list of trials
nct_ids = []
for trial in all_trials:
    nct_ids.append(trial['protocolSection']['identificationModule']['nctId'])
print(len(nct_ids), len(set(nct_ids)))
```

```python
#Save the list of trials to a JSON file with the current date in the filename
date_today = date.today()
os.makedirs('data/raw', exist_ok=True)
with open(f'data/raw/t1d_phase3_trials{date_today}_.json', 'w') as f:
    json.dump(all_trials, f, indent=2)
```
